```python
for i in range(5):
    for j in range(i+1):
        print("*", end="")
    print()
```

```python
for i in range(5):  # Loop for rows
    for j in range(i + 1):  # Loop for columns
        print((i + j) % 2, end=" ")  # Print alternating pattern
    print()  # Move to the next line
```

```python
for i in range(4):
    for j in range (i+1):
        print(j+1, end=" ")
    print()
```

```python
for i in range(5):
    for j in range(i+1):
        print(chr(65 + j), end=" ")
    print()
```

```python
num = 1
for i in range(5):
    for j in range(i+1):
        print(num, end=" ")
        num += 1
    print()
```

```python
for i in range(5):
    for j in range(i+1):
        print(chr(65 + i), end=" ")
    print()
```

```python
for i in range(5):
    for j in range(5-i):
        print(chr(65 + i), end=" ")
    print()
```
