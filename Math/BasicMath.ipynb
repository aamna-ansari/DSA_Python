# **Count digits in a number**

**Example 1:**

Input:N = 12345

Output:5

Explanation:  The number 12345 has 5 digits.

```python
def count_digits(n):
    count = 0
    while n > 0:
        count += 1
        n //= 10
    return count

# Example usage
N = 12345
print("Output:",count_digits(N))

# Built in function 
def countDigit():
    print("Output:",len(str(N)))
countDigit()
```
