# **Recursion**
when a function call itself until a special condition is met.


```python
def main(n):
    if n <= 0:  # Base case: Stop when n reaches 0
        return
    print(n)
    main(n - 1)  

if __name__ == "__main__":
    main(10)  
```

```python
def main(n, limit):
    if n > limit:  # Base case
        return
    print(n)
    main(n + 1, limit)  

if __name__ == "__main__":
    main(1, 10)  

```

 # **Sum of first N numbers**

```python
# functional Recursion
def printSum(n):
    if n == 0:
        return 0
    return n + printSum(n-1)

printSum(5)
```

```python
def printSum(n):
    if n == 0:
        return 0
    return n**3 + printSum(n-1)

printSum(5)
```

```python
# paramenter Recursion
def printSum(n, sum ):
    if n == 0:
        return sum
    return printSum(n-1, sum+n)

printSum(5, 0)
```

# **Factorial of N numbers**

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

factorial(4)
```

# **Palindrome**

```python
string = "A man, a plan, a canal: Panama"

def isPalindrome(string, i):
    string = ''.join(filter(str.isalnum, string)).lower()  # Normalize the string
    if i >= len(string) // 2:
        return True
    if string[i] != string[len(string) - i - 1]:
        return False
    return isPalindrome(string, i + 1)
print(isPalindrome(string, 0))  # Output: True
```
