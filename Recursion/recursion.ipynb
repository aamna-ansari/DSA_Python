# **Recursion**
when a function call itself until a special condition is met.


```python
def main(n):
    if n <= 0:  # Base case: Stop when n reaches 0
        return
    print("Hello, World!")
    main(n - 1)  # Recursive call with a decremented value

if __name__ == "__main__":
    main(5)  # Call with the number of times to print
```
