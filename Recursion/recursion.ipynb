# **Recursion**
when a function call itself until a special condition is met.


```python
def main(n):
    if n <= 0:  # Base case: Stop when n reaches 0
        return
    print(n)
    main(n - 1)  

if __name__ == "__main__":
    main(10)  
```

```python
def main(n, limit):
    if n > limit:  # Base case
        return
    print(n)
    main(n + 1, limit)  

if __name__ == "__main__":
    main(1, 10)  

```

```python
def printNos(n):
        if n==0:
            return 
        else:
            printNos(n-1)
            print(n)
n = 10            
printNos(n)
```
